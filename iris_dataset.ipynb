## comparison of two model on iris dataset

```python
#import pandas library for data manipulation
import pandas as pd 
```

```python
#load iris dataset from sklearn
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
# x is a features of iris data
x=iris.data
# y is a dependent variable of x
y=iris.target
print(y)
```

```python
# import train_test_split for convert data into for test purpose
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.4)
```

```python
print(x.shape)
print(y.shape)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# import kneighbor model for prediction
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
```

```python
knn.fit(x_train,y_train)
```

```python
y_pre=knn.predict(x_test)
```

```python
print(y_pre)
```

```python
# import accuracy_score to cheack model accuracy 
from sklearn.metrics import accuracy_score
```

```python
# accuracy of knn model
print(accuracy_score(y_pre,y_test))
```

```python
#import logistic model for prediction
from sklearn.linear_model import LogisticRegression
```

```python
logreg=LogisticRegression()
```

```python
logreg.fit(x_train,y_train)
```

```python
y_pred=logreg.predict(x_test)
```

```python
print(y_pred)
```

```python
# accuracy_score of logistic model
print(accuracy_score(y_pred,y_test))
```
